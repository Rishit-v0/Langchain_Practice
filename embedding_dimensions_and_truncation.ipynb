```python
from openai import OpenAI
from dotenv import load_dotenv
import numpy as np

load_dotenv()
client = OpenAI()
```

```python
def embed(text: str, dimensions: int = None) -> list[float]:
    kwargs = {
        "model": "text-embedding-3-small",
        "input": text,
    }

    if dimensions:
        kwargs["dimensions"] = dimensions

    response = client.embeddings.create(**kwargs)
    return response.data[0].embedding
```

```python
full = embed("PostgreSQL composite indexes for multi-tenant queries")
small = embed("PostgreSQL composite indexes for multi-tenant queries", dimensions=256)

print(f"Full embedding: {len(full)} dimensions")
print(f"Truncated: {len(small)} dimensions")
```
