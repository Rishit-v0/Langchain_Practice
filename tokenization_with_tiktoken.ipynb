```python
# You can count tokens before sending using tiktoken
import tiktoken

enc = tiktoken.encoding_for_model("gpt-5-nano")
tokens = enc.encode("Hello, how many tokens is this?")
print(len(tokens))   # 8 tokens
```

```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-5-nano")
```

```python
text = "FastApi is a modern Python web framework for building APIs with high performance and ease of use."
tokens = enc.encode(text)
print(f"Text: {text}")
print(f"Number of tokens: {len(tokens)}")
print(f"Tokens: {tokens}")
```

```python
decoded = enc.decode(tokens)
print(f"Decoded text: {decoded}")
```

```python
token_strings = [enc.decode([t]) for t in tokens]
print(f"Token strings: {token_strings}")
```

```python
examples = [
    "Hello world",                    # simple English
    "def fibonacci(n): return n",     # Python code
    "PostgreSQL composite index",     # technical terms
    "私はエンジニアです",                # Japanese (tokenises differently)
]

for ex in examples:
    tokens = len(enc.encode(ex))
    print(f"{tokens:3d} tokens: {ex}")
```
