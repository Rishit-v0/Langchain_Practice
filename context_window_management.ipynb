```python
from openai import OpenAI
from dotenv import load_dotenv
import tiktoken

load_dotenv()
client = OpenAI()
```

```python
def count_tokens(messages: list[dict], model: str = "gpt-5-nano") -> int:
    enc = tiktoken.encoding_for_model(model)
    total = 0 
    for message in messages:
        total += 4
        total += len(enc.encode(message['content']))
        total += len(enc.encode(message['role']))
    total += 2
    return total
```

```python
def safe_chat(messages: list[dict], max_context: int = 12000) -> str:
    while count_tokens(messages) > max_context and len(messages) > 2:
        messages.pop(1)
    
    response = client.chat.completions.create(
        model="gpt-5-nano",
        messages=messages,
    )

    return response.choices[0].message.content
```

```python
messages = [
    {"role": "system", "content": "You are a backend engineering coach."},
    {"role": "user", "content": "What is connection pooling?"},
]

print(f"Token count before sending: {count_tokens(messages)}")
```
